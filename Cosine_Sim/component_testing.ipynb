### Handling Zero Vectors

[0, 0] / 0 -> division by zero issue

- we prevent by replacing 0 with 1 to preserve the zero vector

[0, 0]/ 1 = [0, 0]

```python
#consider 
norm = np.array([
    [5],
    [0],
    [12]
])
print(X / norm)

#Divide by Zero Issue
```

```python
# FIX
print('Before handling zero vectors: ')
norm = np.array([
    [5],
    [0],
    [12]
])
print(norm)

print('After handling zero vectors: ')
safe = np.where(norm>0, norm, 1.0)
print(safe)

print('\n')
print('Complete normalization with zero handling')
print(X / safe)

print('\n')
print('Complete normalization WITHOUT zero handling')
print(X / norm)

```

```python
def normalize_rows(X):
    norm = np.linalg.norm(X,
                          axis=1,
                          keepdims=True)
    #handle zero div edge case
    safe = np.where(norm > 0, norm, 1.0)
    return X / safe
```

## Cosine Similarity Matrix of Query and Document vectors

```python
Q = np.array([
    [1, 0],
    [0, 1]
])
print(Q.shape)
D = np.array([
    [1, 1],
    [1, 0],
    [0, 1]
])
print(D.shape)
```

```python
norm_q = normalize_rows(Q)
print('Normalized Q: '+'\n', norm_q)
print(norm_q.shape)

print('\n')
norm_d = normalize_rows(D)
print('Normalized D: '+'\n', norm_d)
print(norm_d.shape)
```

```python
print(norm_d.T)
```

```python
scores_matrix = norm_q @ norm_d.T
print(scores_matrix)
```

### Argsorting

```python
x = np.asarray([2,3,4])
y = np.asarray([3,1,6])
sorted_x = np.argsort(x)
sorted_y = np.argsort(y)

print(sorted_x)
print(sorted_y)
```

```python
# pretend this is scores matrix bw Q and D documents after matrix multiplication
score_matrix = np.asarray([
    [1,2,3,6],
    [5,1,3,0],
    [2,1,7,9]
])

scored = np.argsort(score_matrix)
print(scored)
```

```python
#in descending order
scored = np.argsort(-score_matrix)
print(scored)
```

```python
#across axis=1
scored = np.argsort(score_matrix, axis=1, kind='stable')
print(scored)
```

Column 0 of order: Each query's #1 best document index (pointing to D[idx]).

Column 1 of order: Each query's #2 best document index (pointing to D[idx]).

```python
print(scored[:, :2])
```

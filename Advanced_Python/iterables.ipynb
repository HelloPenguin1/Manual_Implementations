### Iterator
Object that allows the programmer to traverse through a sequence of data w/o having to store the entire data in memory

*Efficient*

```python
# Traditional 
L = [x for x in range(10000)] # consider any large number
for i in L:
    print(i * 2)
```

```python
import sys
print("Memory Being Used...")
print("...Without an iterator: ", sys.getsizeof(L)/64)

# VERSUS a ITERATOR, Check how less the memory being taken is in comparison
x = range(10000)
print("...With an iterator:", sys.getsizeof(x)/64)
```

### Iterable
**Object** which one can iterate over. [iteration: general term for going over a group of items one after the other], allowing one to use a for-loop over it for instance

Ex: list, tuple, str, dict

Generates an iterator when passed to iter() method


```python
l = [1, 2, 3]
print(type(l))

print(type(iter(l)))

#iter(l) --> iterator,   the method turns an iterable (list) to an ITERATOR , in this case a list iterator
```

### Understanding NEXT function and how for-loop actually works

next(iterator) ->
1. Returns items one by one. Each time you call next(), it yields the current item and moves the pointer to the next one. Once an item is consumed, the iterator cannot go backward.
2. Handles exhaustion. When there are no items left, next() raises a StopIteration exception. If you pass a default argument, it returns that value instead of crashing.

```python
nums = [10, 20]
it = iter(nums)  # Create an iterator

print(next(it))        # Output: 10
print(next(it))        # Output: 20
print(next(it, "End")) # Output: 'End' (avoids StopIteration)
#print(next(it))        # Raises StopIteration error
```

```python
print(next(it)) #without default arg to handle exhaustion -> raises StopIteration exception
```

```python
# Maual implementation of for loop
def my_for_loop(any_iterable):
    iterator = iter(any_iterable)
    while True:
        try:
            print(next(iterator))
        except StopIteration:
            break
```

```python
a = [1,2,3,4]
b = (1,2,3)
c = {4,5,6}
d = {1:2, 2:5, 3:6}

my_for_loop(d)
```

### Core Mechanics

* **Iterator protocol:** A Python object implementing `__iter__()` and `__next__()`, raising `StopIteration` when exhausted.
* **Single-pass consumption:** Once consumed, an iterator or generator cannot be rewound—you must instantiate a new one for another pass.
* **Lazy evaluation:** Values are computed or fetched on demand when `next()` (or a `for` loop) requests them, shifting when errors happen from setup time to iteration time.

### Key AI Engineering Use Cases

* **LLM token streaming:** Consuming chunks incrementally as a model generates responses instead of waiting for the full completion.
* **Data ingestion & batching:** Reading large corpora, JSONL files, or embedding batches in fixed chunks to control RAM and API payload sizes.
* **Agent & graph execution:** Stepping through intermediate tool calls, retrieval stages, or state transitions incrementally.
* **Evaluation & retrieval pipelines:** Iterating over test queries, retrieved documents, or reranked candidates without materializing huge intermediate lists.

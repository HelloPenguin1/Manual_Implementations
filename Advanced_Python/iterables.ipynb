### Iterator
Object that allows the programmer to traverse through a sequence of data w/o having to store the entire data in memory

*Efficient*

```python
# Traditional 
L = [x for x in range(10000)] # consider any large number
for i in L:
    print(i * 2)
```

```python
import sys
print("Memory Being Used...")
print("...Without an iterator: ", sys.getsizeof(L)/64)

# VERSUS a ITERATOR, Check how less the memory being taken is in comparison
x = range(10000)
print("...With an iterator:", sys.getsizeof(x)/64)
```

### Iterable
**Object** which one can iterate over. [iteration: general term for going over a group of items one after the other]

Generates an iterator when passed to iter() method
